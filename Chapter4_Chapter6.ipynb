Exercise 4.1 – Guess the Number

```python
# Exercise 4.1

secret = 7
guess = 5

if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
```

Exercise 4.2 – Small and Green

```python
# Exercise 4.2

small = True
green = True

if small and green:
    print("pea")
elif small and not green:
    print("cherry")
elif not small and green:
    print("watermelon")
else:
    print("pumpkin")
```

Exercise 6.1 – For Loop

```python
# Exercise 6.1

numbers = [3, 2, 1, 0]

for number in numbers:
    print(number)
```

Exercise 6.2 – While Loop

```python
# Exercise 6.2

guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break

    number += 1
```

Exercise 6.3 – For Loop with Range

```python
# Exercise 6.3

guess_me = 5

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
```
